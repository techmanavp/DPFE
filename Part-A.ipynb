## Q1. What is Data Analysis?
**Data analysis** is the process of collecting, cleaning, transforming and examining raw data to discover useful information, patterns and conclusions that support decision making.

Main types:
| Type | Question it answers | Example here |
|---|---|---|
| Descriptive | What happened? | Average orders per customer |
| Diagnostic | Why did it happen? | Why do customers with complaints churn more? |
| Predictive | What will happen? | Which customer will churn next month? |
| Prescriptive | What should we do? | Offer cashback to high-risk customers |

## Q2. Planning a Data Science Project (all steps)
1. **Problem definition:** understand the business goal and success criteria.
2. **Data collection / acquisition:** gather data from CSV, JSON, databases and APIs.
3. **Data understanding:** inspect structure, types, summary statistics.
4. **Data cleaning:** handle missing values, duplicates, wrong data types, inconsistent labels, outliers.
5. **Exploratory Data Analysis (EDA):** univariate, bivariate and multivariate analysis with visualisations.
6. **Feature engineering:** create new features, encode categories, scale numbers.
7. **Model building:** split data, choose algorithms, train.
8. **Model evaluation:** accuracy, precision, recall, F1, ROC-AUC; tune the model.
9. **Deployment:** put the model into production (API / dashboard).
10. **Monitoring & maintenance:** track performance and retrain when data changes.

## Q3. Machine Learning Problem Statement
| Item | Definition |
|---|---|
| **Business problem** | Customers are leaving the platform (churn), which reduces revenue. Acquiring a new customer costs more than retaining one. |
| **ML task** | Supervised learning, **binary classification** |

## Q4. Tensors

**What is a tensor?** A tensor is a multi-dimensional array of numbers. It generalises scalars, vectors and matrices to any number of dimensions (axes). Deep learning libraries such as TensorFlow and PyTorch store all data and model weights as tensors.

| Rank (ndim) | Name | Example |
|---|---|---|
| 0 | Scalar | a single number, `5` |
| 1 | Vector | one customer's feature list |
| 2 | Matrix | our whole dataset (rows x columns) |
| 3 | 3D tensor | one colour image (height x width x channels) |
| 4 | 4D tensor | a batch of images (batch x height x width x channels) |

Key properties: **ndim** (number of axes), **shape** (size of each axis), **size** (total elements), **dtype** (data type).

```python
import numpy as np

scalar = np.array(5)                                   # 0D
vector = np.array([35, 3, 160])                        # 1D: [age, orders, cashback] of one customer
matrix = np.array([[35, 3, 160],
                   [28, 1, 120],
                   [41, 7, 210]])                      # 2D: 3 customers x 3 features
tensor3d = np.random.rand(4, 5, 3)                     # 3D: e.g. a tiny 4x5 RGB image
tensor4d = np.random.rand(32, 28, 28, 3)               # 4D: batch of 32 RGB images of 28x28

for name, t in [("Scalar", scalar), ("Vector", vector), ("Matrix", matrix),
                ("3D tensor", tensor3d), ("4D tensor", tensor4d)]:
    print(f"{name:10s} | ndim = {t.ndim} | shape = {t.shape} | size = {t.size} | dtype = {t.dtype}")
```

```python
# Common tensor operations
print("Matrix:\n", matrix)
print("\nReshape (1x9):", matrix.reshape(1, 9))
print("\nTranspose (3 features x 3 customers):\n", matrix.T)
print("\nElement-wise x2:\n", matrix * 2)
print("\nMatrix multiplication (matrix @ matrix.T):\n", matrix @ matrix.T)
print("\nIndexing  matrix[0, 1] =", matrix[0, 1], "(first customer, 2nd feature)")
print("Slicing   matrix[:, 2] =", matrix[:, 2], "(cashback of all customers)")
print("Mean of each feature (axis=0):", matrix.mean(axis=0))
print("Mean of each customer (axis=1):", matrix.mean(axis=1))

# Broadcasting: subtract each column's mean (standard step before scaling)
centered = matrix - matrix.mean(axis=0)
print("\nMean-centred matrix (broadcasting):\n", centered)
```
