## Task 1 - Short Notes

### 1.1 What is Data Analysis?

**Data analysis** is the process of collecting, cleaning, transforming and examining data to discover useful
information, draw conclusions and support decision making.

**Main steps**

1. **Define the question** - what do we want to know? (e.g. *Which customers are likely to default?*)
2. **Collect data** - CSV, JSON, SQL databases, APIs, web, sensors.
3. **Clean data** - fix missing values, duplicates, wrong types, outliers.
4. **Explore (EDA)** - summary statistics, distributions, correlations, visualisation.
5. **Model / analyse** - apply statistics or machine learning.
6. **Interpret and communicate** - charts, dashboards, reports, recommendations.

**Types of data analysis**

| Type | Question it answers | Example in this project |
|---|---|---|
| Descriptive | What happened? | Average income, % of defaulters |
| Diagnostic | Why did it happen? | Why do low credit-score customers default more? |
| Predictive | What will happen? | Predict `default_flag` for a new customer |
| Prescriptive | What should we do? | Reject / approve / reduce loan amount |

> Data preprocessing is the part of data analysis that takes about 60-80 % of a data scientist's time,
> because *"garbage in = garbage out"*.

### 1.2 How to Plan a Data Science Project

A common framework is **CRISP-DM** (Cross Industry Standard Process for Data Mining):

| Phase | What we do | In this project |
|---|---|---|
| 1. Business understanding | Define goal, success metric, constraints | Reduce loan defaults for a fintech company |
| 2. Data understanding | Collect data, explore quality | Load CSV + JSON + SQL + API, profile data |
| 3. Data preparation | Clean, impute, encode, scale, engineer features | Parts C to G |
| 4. Modelling | Choose algorithms, train, tune | (next step - out of scope) |
| 5. Evaluation | Check metrics against the business goal | Recall, Precision, F1, PR-AUC |
| 6. Deployment | Put in production, monitor | Save the preprocessing pipeline (`.joblib`) |

**Planning checklist**

- Clear problem statement and measurable success criteria
- Data sources, owners, access and privacy (customer data is sensitive)
- Timeline and milestones (this project: 8 days x 1 hour)
- Tools (Python, pandas, scikit-learn, Git/GitHub)
- Risks (class imbalance, missing data, data leakage, bias)
- Reproducibility (fixed random seed, version control, documented notebooks)

### 1.3 How to Frame a Machine Learning Problem

Framing means converting a **business problem** into a **machine-learning task**.

| Question | Answer for this project |
|---|---|
| Business problem | Company loses money on customers who do not repay loans |
| ML task | **Supervised, binary classification** |
| Target (label) | `default_flag` (0 = No default, 1 = Default) |
| Input features | Demographics, financial details, behavioural attributes |
| Data available at prediction time? | Yes - all features are known when the loan is applied for (avoid leakage) |
| Evaluation metric | The data is imbalanced (about 20 % defaulters) so use **Recall, Precision, F1, ROC-AUC, PR-AUC**, not just accuracy |
| Baseline | Predict "no default" for everyone (about 80 % accuracy but useless) |
| Cost of errors | A missed defaulter (false negative) costs more than a rejected good customer (false positive) |
| Constraints | Explainability, fairness (gender, region), data privacy |

**Types of ML problems:** classification, regression, clustering, recommendation, anomaly detection.

## Task 2 - Tensors (in-depth, with NumPy)

A **tensor** is a multi-dimensional array of numbers that all share the same data type.
It is the basic data structure of NumPy, TensorFlow and PyTorch.

| Rank (ndim) | Name | Example | Shape |
|---|---|---|---|
| 0 | Scalar | a single credit score `720` | `()` |
| 1 | Vector | incomes of 5 customers | `(5,)` |
| 2 | Matrix | table: customers x features | `(1000, 12)` |
| 3 | 3-D tensor | time series: customers x months x features | `(1000, 6, 3)` |
| 4 | 4-D tensor | batch of colour images: batch x H x W x channels | `(32, 64, 64, 3)` |

**Key attributes:** `ndim` (number of axes), `shape` (size of each axis), `size` (total elements), `dtype`.

```python
import numpy as np

# ---- 1. Creating tensors of different ranks
scalar = np.array(720)                                  # rank 0
vector = np.array([450000, 720000, 380000, 910000, 560000])   # rank 1
matrix = np.array([[25, 700, 1],
                   [41, 640, 0],
                   [33, 590, 1]])                       # rank 2
tensor3d = np.arange(24).reshape(2, 3, 4)               # rank 3
tensor4d = np.zeros((2, 4, 4, 3))                       # rank 4 (2 tiny RGB images 4x4)

for name, t in [("scalar", scalar), ("vector", vector), ("matrix", matrix),
                ("3D tensor", tensor3d), ("4D tensor", tensor4d)]:
    print(f"{name:10s} ndim={t.ndim}  shape={t.shape}  size={t.size}  dtype={t.dtype}")
```

```python
# ---- 2. Other ways to create tensors
print(np.zeros((2, 3)))
print(np.ones((2, 2)))
print(np.full((2, 2), 7))
print(np.eye(3))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
rng = np.random.default_rng(42)
print(rng.normal(loc=0, scale=1, size=(2, 3)).round(2))
```

```python
# ---- 3. Indexing and slicing
t = np.arange(24).reshape(2, 3, 4)
print("t[0, 1, 2]     ->", t[0, 1, 2])        # one element
print("t[1]           ->\n", t[1])            # a whole 2-D slice
print("t[:, 0, :]     ->\n", t[:, 0, :])      # first row of each matrix
print("t[..., -1]     ->\n", t[..., -1])      # last element along the last axis

# boolean masking (very common in preprocessing)
credit = np.array([650, 720, 580, 810, 300, 705])
print("credit > 700   ->", credit > 700)
print("credit[credit > 700] ->", credit[credit > 700])
```

```python
# ---- 4. Reshape, flatten, transpose, expand dims
a = np.arange(12)
print("reshape (3,4):\n", a.reshape(3, 4))
print("reshape (2,-1) -> shape", a.reshape(2, -1).shape)   # -1 = infer
m = a.reshape(3, 4)
print("transpose shape:", m.T.shape)
print("flatten:", m.flatten())
print("expand_dims:", np.expand_dims(a, axis=0).shape, np.expand_dims(a, axis=1).shape)
print("swapaxes on 3D:", np.arange(24).reshape(2, 3, 4).swapaxes(0, 2).shape)
```

```python
# ---- 5. Element-wise operations, aggregations and axes
X = np.array([[25, 450000., 700.],
              [41, 720000., 640.],
              [33, 380000., 590.],
              [52, 910000., 760.]])        # 4 customers x 3 features (age, income, credit score)

print("X + 10 :\n", X + 10)
print("X * 2  :\n", X * 2)
print("column means (axis=0):", X.mean(axis=0))
print("row sums    (axis=1):", X.sum(axis=1))
print("max per column      :", X.max(axis=0))
print("argmax per column   :", X.argmax(axis=0))
print("std per column      :", X.std(axis=0).round(2))
```

```python
# ---- 6. Broadcasting: automatic expansion of smaller tensors
# Standardisation (z-score) of every column in ONE line, no loop:
mean = X.mean(axis=0)          # shape (3,)
std = X.std(axis=0)            # shape (3,)
Z = (X - mean) / std           # (4,3) - (3,) -> broadcast over rows
print("mean of Z:", Z.mean(axis=0).round(6))
print("std  of Z:", Z.std(axis=0).round(6))

# Broadcasting rule: compare shapes from the right; dims must be equal or 1.
a = np.ones((3, 1))
b = np.arange(4)
print((a + b).shape)           # (3,1) + (4,) -> (3,4)
```

```python
# ---- 7. Tensor algebra: dot, matmul, tensordot, einsum
w = np.array([0.001, 0.0000005, -0.01])      # toy weight vector (3,)
print("X @ w (matrix-vector) :", (X @ w).round(3))

A = rng.integers(1, 5, (2, 3))
B = rng.integers(1, 5, (3, 2))
print("A @ B shape:", (A @ B).shape)
print("np.dot == matmul for 2-D:", np.array_equal(A @ B, np.dot(A, B)))
print("einsum matmul:", np.array_equal(np.einsum("ij,jk->ik", A, B), A @ B))
print("tensordot:", np.tensordot(A, B, axes=1).shape)
```

```python
# ---- 8. Joining and splitting tensors
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print("vstack:\n", np.vstack([a, b]))
print("hstack:\n", np.hstack([a, b]))
print("stack (new axis) shape:", np.stack([a, b]).shape)
print("split:", [s.shape for s in np.split(np.arange(12).reshape(6, 2), 3)])
```

```python
# ---- 9. Memory: views vs copies
base = np.arange(6)
view = base[:3]            # a VIEW - shares memory
view[0] = 99
print("base after changing view:", base)       # base changed too!
safe = base[:3].copy()     # an independent COPY
safe[0] = -1
print("base after changing copy:", base)       # unchanged
print("dtype & memory:", X.dtype, X.nbytes, "bytes")
print("astype float32:", X.astype(np.float32).nbytes, "bytes")
```

### Where tensors appear in this project

| Object | Tensor rank / shape |
|---|---|
| Final feature table `X` | rank 2 - `(n_customers, n_features)` |
| Target `y` (`default_flag`) | rank 1 - `(n_customers,)` |
| Scaling / standardisation | broadcasting of a `(n_features,)` vector over the matrix |
| Distance in KNN imputer | vector arithmetic on rows |
| A deep-learning model later | a batch of customers = a rank-2 or rank-3 tensor |

**Summary:** a tensor generalises scalar, vector and matrix. NumPy's `ndarray` lets us do fast, vectorised
maths with **shape**, **axes**, **broadcasting** and **linear algebra**, which is what every preprocessing step
in the next notebook is built on.
